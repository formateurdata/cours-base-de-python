## 4. Calculer plutôt que recopier un résultat

Sans guillemets, Python peut interpréter des nombres et des opérations. `+` additionne, `*` multiplie et `/` effectue une division. Les parenthèses permettent de choisir l’ordre du calcul.

La première station possède 5 vélos mécaniques et 2 électriques dans l’extrait. Vérifions le total.

````python
print(5 + 2)
print(7 / 17)
print((5 + 2) * 2)
````

Le total calculé vaut 7. La division donne une proportion par rapport aux 17 places de la station. Cette proportion ne permet pas de déduire les bornes libres : certaines places peuvent être indisponibles.

## 5. Nommer une valeur avec une variable

Une variable permet de retrouver une valeur par un nom. Dans `velos = 7`, le signe `=` affecte la valeur de droite au nom de gauche. Il ne pose pas une question d’égalité. Choisissez des noms qui expriment ce que représente la valeur.

Une ligne commençant par `#` est un commentaire : elle explique le code sans être exécutée.

````python
# Disponibilité observée dans notre extrait local.
velos = 7
besoin = 3
reste = velos - besoin
print("Vélos restants après le départ du groupe :", reste)
````

Le résultat utilise les valeurs mémorisées dans `velos` et `besoin`. Modifiez seulement `besoin`, puis relancez la cellule entière. Un nom de variable est sensible aux majuscules : `Velos` et `velos` sont différents.

## 6. Quotient entier et reste

L’opérateur `//` donne le quotient entier de la division ; `%` donne son reste. Pour nos durées positives, ils permettent de séparer les heures et les minutes.

Cet exemple part de 95 minutes : observez les deux résultats.

````python
duree = 95
heures = duree // 60
minutes = duree % 60
print(heures, "h", minutes, "min")
````

L’affichage est `1 h 35 min`. Une valeur numérique s’écrit sans guillemets. `"95"` représente du texte ; nous allons maintenant utiliser les conversions.

## 7. Une valeur a un type

Le type décrit les opérations possibles sur une valeur. Nous utiliserons `int` pour un entier, `float` pour un nombre à virgule, `str` pour du texte et `bool` pour un booléen (`True` ou `False`). En code Python, la partie décimale s’écrit avec un point.

Le code d’une station reste un `str` : c’est une référence à conserver, pas une quantité à additionner.

````python
code_station = "1001"
velos = 7
taux = velos / 17
est_disponible = velos > 0
print(type(code_station), type(velos))
print(type(taux), type(est_disponible))
````

`type(...)` affiche les quatre types. La comparaison `velos > 0` produit ici `True`, que nous pourrons utiliser dans une condition.

## 8. Lire une saisie puis la convertir

`input(...)` affiche une question et renvoie la réponse sous forme de texte. `int(...)` transforme une chaîne représentant un entier en entier ; `float(...)` accepte une valeur décimale écrite avec un point.

À l’exécution, saisissez **3** pour cet exemple. Les sorties enregistrées ont été obtenues avec cette saisie.

````python
saisie = input("Combien de vélos demandez-vous ? ")
print("Type reçu :", type(saisie))
besoin = int(saisie)
print("Vélos restants :", 7 - besoin)
````

Même si l’on tape des chiffres, la saisie est du texte avant conversion. Taper `trois` provoquerait une `ValueError` : nous apprendrons à traiter cette situation en séance 6. Pour l’instant, respectez le type demandé.

## 9. Présenter un résultat calculé

Une chaîne préfixée par `f` insère les valeurs entre accolades. `:.2f` demande deux chiffres après le point pour cet affichage. Cela ne transforme pas la variable en montant arrondi pour tous les calculs.

Saisissez **2** dans l’exemple. Le prix est une valeur inventée pour apprendre à calculer un devis.

````python
quantite = int(input("Quantité entière : "))
prix_unitaire = 12.5
montant = quantite * prix_unitaire
print(f"Pour {quantite} articles : {montant:.2f} euros")
````

Pour une quantité de 2, le montant affiché est `25.00 euros`. Avant d’exécuter un programme interactif, vérifiez toujours les types de saisie annoncés.

## 10. Comparer des valeurs et combiner des conditions

Un booléen représente `True` ou `False`. Les opérateurs `==`, `!=`, `<`, `<=`, `>` et `>=` comparent des valeurs. `and` demande deux conditions vraies ; `or` en demande au moins une ; `not` inverse un booléen. Distinguez `=` pour affecter et `==` pour comparer.

Le besoin doit être positif et ne pas dépasser la disponibilité de la station.

````python
velos = 7
besoin = 3
print(velos >= besoin)
print(besoin > 0 and velos >= besoin)
print(besoin == velos)
````

Les deux premières comparaisons donnent True, la dernière False. Essayez les besoins 0, 7 et 8 : les valeurs de frontière permettent de vérifier exactement la règle.

## 11. Choisir avec une condition

`if` exécute son bloc si sa condition vaut `True`. `elif` propose un autre cas ; `else` couvre les cas restants. Le premier bloc dont la condition est vraie est exécuté. Le deux-points annonce le bloc ; les lignes du bloc sont indentées de quatre espaces.

Pour la station qui possède 7 vélos, comparons un besoin de 3 vélos à cette disponibilité.

````python
velos = 7
besoin = 3
if besoin <= 0:
    print("Le besoin doit être positif.")
elif velos >= besoin:
    print("Le groupe peut partir.")
else:
    print("Il manque des vélos.")
````

Le programme exécute un seul des trois blocs. Essayez ensuite un besoin de 7, de 8 et de 0 : ces manipulations vérifient la limite et le cas invalide.

## 12. Affecter, réaffecter et convertir

Python évalue la partie droite avant d’affecter la nouvelle valeur. Un même nom peut donc recevoir le résultat d’un calcul qui utilise son ancienne valeur. Le type dépend de la valeur reçue ; une conversion doit être volontaire.

Comparez l’addition de nombres et la concaténation de textes. Le résultat visuellement proche ne signifie pas que les opérations sont identiques.

````python
quantite = 2
quantite = quantite + 1
print("Quantité :", quantite)
print("2" + "3")
print(int("2") + int("3"))
print(10 / 4, 10 // 4, 10 % 4)
````

La quantité devient 3. Les textes donnent `23` ; les nombres donnent 5. `/`, `//` et `%` répondent à trois besoins distincts. Un texte non convertible donnera une erreur, étudiée plus tard.

## 13. Contrôler les unités d’un calcul

Avant de programmer, nommez les données, leur unité et le résultat attendu. Une durée en minutes et un prix en euros n’ont pas le même sens, même si les deux sont stockés comme nombres. Une variable ne remplace pas ce raisonnement.

Un nom avec une unité rend aussi l’erreur plus facile à repérer.

````python
duree_minutes = 90
prix_horaire_euros = 12
montant_euros = duree_minutes / 60 * prix_horaire_euros
print(f"Durée : {duree_minutes} minutes")
print(f"Montant : {montant_euros:.2f} euros")
````

Le montant est 18.00 euros dans cette situation inventée. Multiplier directement 90 par 12 aurait mélangé les minutes et les heures.
