## 1. Parcourir des valeurs avec `for`

Une boucle `for` exécute le même bloc pour chaque élément d’une collection. Les valeurs entre crochets sont fournies ici ; nous apprendrons à construire et modifier ces listes à la séance 4.

À chaque tour, `velos` reçoit la valeur suivante. Observez le compteur, qui est initialisé **avant** la boucle.

```python
valeurs_velos = [7, 14, 21, 14, 6, 22, 15, 45]
compteur = 0
for velos in valeurs_velos:
    if velos >= 10:
        compteur = compteur + 1
print("Stations avec au moins 10 vélos :", compteur)
```

On obtient 6 stations sur les 8 de l’extrait. Initialiser le compteur à l’intérieur de la boucle le remettrait à zéro à chaque tour. L’indentation de `print()` le place ici après le parcours.

## 2. Répéter un nombre connu de fois

`range(debut, fin)` fournit des entiers du début inclus à la fin exclue. Pour afficher les nombres de 1 à 3, la borne de fin est donc 4. L’écriture `range(3)` commencerait à 0.

Prédisez les trois lignes avant d’exécuter.

```python
for numero in range(1, 4):
    print("Tour", numero)
```

Le dernier nombre affiché est 3. Lorsque vous utilisez une borne, testez le premier et le dernier éléments attendus pour repérer un décalage d’un tour.

## 3. Répéter tant qu’une condition est vraie

`while` vérifie sa condition avant chaque tour. Cette boucle convient quand le nombre de tours dépend de ce qui se passe. Une variable doit évoluer pour permettre à la condition de devenir fausse.

Ici, la réserve n’est qu’une simulation de départs successifs ; elle ne modifie pas les données Vélib’.

```python
reserve = 7
while reserve > 0:
    print("Réserve avant départ :", reserve)
    reserve = reserve - 1
print("Fin de la simulation")
```

La réserve diminue de 7 à 0 ; le dernier tour commence avec 1. Oublier la décrémentation provoquerait une boucle sans fin. Pour arrêter une cellule, utilisez le bouton carré d’interruption du noyau.

## 4. Réutiliser un module existant

`import random` rend disponible le module `random`. `random.randint(1, 20)` tire un entier entre 1 et 20, **deux bornes incluses**. Le point permet de désigner une fonction du module.

Nous fixons une graine pour retrouver le même exemple pendant la préparation. Pour un jeu dont on souhaite renouveler les tirages, on ne réinitialise pas cette graine à chaque partie.

```python
import random
random.seed(42)
nombre_secret = random.randint(1, 20)
print("Tirage de démonstration :", nombre_secret)
```

Avec cet exemple sous Python 3.12, le tirage affiché est 4. Le jeu d’application utilise d’abord un secret fixé par la consigne afin de vérifier ses branches de manière prévisible.

## 5. Lire `break` et `continue`

`continue` passe au tour suivant de la boucle. `break` quitte la boucle qui le contient. Ces deux instructions peuvent clarifier certains parcours ; elles ne remplacent pas la compréhension de la condition d’arrêt.

Lisez d’abord les conditions, puis prédisez les valeurs affichées.

```python
for nombre in range(1, 6):
    if nombre == 2:
        continue
    if nombre == 4:
        break
    print(nombre)
```

Le programme affiche 1 et 3 : 2 est sauté, puis la boucle s’arrête avant d’afficher 4. Une boucle `while` avec une condition explicite suffit pour la devinette demandée.
