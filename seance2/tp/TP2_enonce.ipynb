## 3. Des branches exclusives et des conditions ordonnées

`if / elif / else` choisit le premier bloc dont la condition est vraie. L’ordre compte si plusieurs conditions pourraient être vraies. Ici, un besoin invalide doit être refusé avant de comparer la quantité disponible.

Le deux-points et l’indentation délimitent les blocs. Le dernier affichage, non indenté, est exécuté après la décision.

```python
velos = 7
besoin = 7
if besoin <= 0:
    print("Besoin invalide")
elif besoin <= velos:
    print("Départ possible")
else:
    print("Vélos insuffisants")
print("Décision terminée")
```

Le besoin égal à 7 est accepté. Le remplacer par 0 ou 8 change la branche. Deux `if` indépendants pourraient exécuter deux blocs ; une chaîne `if/elif` exprime ici un seul choix.

## 4. Combiner des règles sans les confondre

`and` exige deux conditions vraies ; `or` en exige au moins une. Avec `or`, écrire « besoin positif ou inférieur à la disponibilité » accepterait des besoins négatifs. Utilisez des parenthèses pour rendre une combinaison plus facile à lire.

La règle cherchée est : besoin strictement positif **et** besoin au plus égal au nombre de vélos.

```python
velos = 7
for besoin in [-1, 0, 7, 8]:
    accepte = (besoin > 0) and (besoin <= velos)
    print("Besoin :", besoin, "| accepté :", accepte)
```

La ligne `for` applique le même test aux quatre valeurs fournies. Seul le besoin 7 est accepté dans ce jeu de vérification. Nous étudions maintenant ce parcours.

## 5. Répéter avec for et range

Une boucle `for` exécute son bloc pour chaque élément reçu. `range(debut, fin)` fournit des entiers : le début est inclus et la fin exclue. Avec un seul argument, le début vaut zéro. Le pas est facultatif.

Choisissez une boucle `for` lorsque vous parcourez des valeurs déjà disponibles ou un nombre de répétitions connu.

```python
for numero in range(1, 4):
    print("Tour :", numero)
print("Nombres pairs :")
for nombre in range(0, 7, 2):
    print(nombre)
```

Le premier parcours affiche 1, 2, 3 ; le second 0, 2, 4, 6. Pour dix multiplicateurs de 1 à 10, la borne finale sera 11. La variable de boucle reçoit sa valeur à chaque tour.

## 6. Parcourir les observations et compter

Un compteur répond à la question « combien ? ». Il est initialisé avant le parcours et augmenté quand une observation satisfait le critère. L’écriture `compteur += 1` abrège `compteur = compteur + 1`.

Le `print` après la boucle est volontairement aligné avec `for` : il affiche un bilan unique.

```python
valeurs_velos = [7, 14, 21, 14, 6, 22, 15, 45]
compteur = 0
for velos in valeurs_velos:
    if velos >= 10:
        compteur += 1
print("Stations avec au moins 10 vélos dans l’extrait :", compteur)
```

Six stations sur huit satisfont le seuil de 10. Un compteur remis à zéro dans la boucle perdrait les tours précédents. Indenter le dernier `print` afficherait des bilans intermédiaires.

## 7. Cumuler puis calculer une moyenne

Un accumulateur additionne des contributions. Un compteur compte des observations. Leur quotient donne une moyenne lorsque l’effectif est non nul.

Le contrôle de l’effectif précède la division. Un résultat vide est une absence d’observations ; on ne lui invente pas une moyenne observée égale à zéro.

```python
nombre = 0
total = 0
for velos in valeurs_velos:
    nombre += 1
    total += velos
if nombre > 0:
    print("Stations :", nombre)
    print("Total de vélos :", total)
    print("Moyenne par station :", total / nombre)
else:
    print("Aucune observation : moyenne non calculée")
```

Le bilan donne 8 stations, 144 vélos disponibles et une moyenne de 18.0 par station. Le total et la moyenne décrivent des mesures différentes, sur le même périmètre.

## 8. Suivre l’état d’un programme

Pour diagnostiquer une boucle, suivez ses variables après chaque tour : valeur reçue, compteur et total. Une trace permet de localiser le premier tour où le programme s’écarte de ce qui était attendu.

Cette petite trace utilise les trois premières disponibilités. Elle n’est pas un nouvel export de données.

```python
compteur = 0
total = 0
for velos in [7, 14, 21]:
    compteur += 1
    total += velos
    print(f"Valeur={velos} | compteur={compteur} | total={total}")
```

Les totaux successifs sont 7, 21 et 42. Si l’on plaçait `total = 0` à l’intérieur du bloc, les valeurs finales ne représenteraient plus un cumul.

## 9. Chercher le maximum sans le connaître à l’avance

Pour une liste non vide, on peut partir de la première observation, puis conserver la meilleure valeur rencontrée. Initialiser arbitrairement le maximum à zéro échouerait sur une série entièrement négative.

`len(...)` donne ici le nombre d’observations et `[0]` lit la première. Ce sont les seules opérations sur la liste nécessaires à cette recherche.

```python
if len(valeurs_velos) == 0:
    print("Maximum absent")
else:
    maximum = valeurs_velos[0]
    for velos in valeurs_velos:
        if velos > maximum:
            maximum = velos
    print("Maximum de l’extrait :", maximum)
```

Le maximum est 45. Il est calculé depuis les observations, pas recopié comme une réponse fixe. Le cas vide est traité avant d’accéder au premier élément.

## 10. Répéter tant qu’une condition est vraie

`while` évalue sa condition **avant** chaque tour. Le corps peut donc ne jamais être exécuté. Il faut faire évoluer l’état pour rendre possible l’arrêt.

La réserve suivante est une simulation de départs, initialisée avec une disponibilité réelle. Elle ne modifie pas les observations de la source.

```python
reserve_simulee = valeurs_velos[0]
while reserve_simulee > 0:
    print("Réserve avant départ :", reserve_simulee)
    reserve_simulee -= 1
print("Simulation terminée")
```

Sept tours sont exécutés, puis la réserve vaut zéro. Retirer la mise à jour rendrait la condition vraie indéfiniment. En cas de boucle sans fin, interrompez la cellule avec le bouton carré du notebook.

## 11. Redemander une valeur hors du domaine autorisé

Le nombre de saisies n’est pas connu à l’avance : `while` convient pour redemander un entier positif. Dans cet exemple, les réponses sont supposées numériques ; le traitement des textes non numériques sera étudié plus tard.

Pour reproduire la sortie enregistrée, saisissez successivement **0**, **-2**, puis **3**. La première saisie existe avant la condition ; la suivante est demandée dans le bloc de refus.

```python
quantite = int(input("Quantité entière positive : "))
while quantite <= 0:
    print("Cette quantité doit être positive.")
    quantite = int(input("Nouvelle quantité : "))
print("Quantité acceptée :", quantite)
```

Deux valeurs sont refusées avant l’acceptation de 3. Si la seconde saisie était oubliée, le programme répéterait le refus sans pouvoir changer d’état.

## 12. Quitter un parcours ou sauter un tour

`break` quitte la boucle courante. `continue` passe au tour suivant sans exécuter le reste du bloc. Ces instructions ne quittent pas automatiquement tout le programme.

Le premier parcours cherche la première observation atteignant 20 vélos, selon l’ordre de l’extrait. Le second ne totalise que les observations atteignant 10 vélos.

```python
for velos in valeurs_velos:
    if velos >= 20:
        print("Première disponibilité au seuil :", velos)
        break
total_selection = 0
for velos in valeurs_velos:
    if velos < 10:
        continue
    total_selection += velos
print("Total des stations retenues au seuil de 10 :", total_selection)
```

La première valeur trouvée est 21 ; ce n’est pas nécessairement le maximum. Le cumul du second parcours vaut 131. Chercher la première réponse, la meilleure réponse et toutes les réponses sont trois besoins distincts.

## 13. Un module pour préparer un jeu

`import random` rend disponible le module standard `random`. Le point dans `random.randint(...)` désigne une fonction du module. Les deux bornes de `randint(1, 20)` sont incluses, contrairement à la borne de fin de `range`.

La graine fixe sert seulement à rendre cette démonstration reproductible pendant la préparation. On ne la réinitialise pas à chaque partie si l’on souhaite renouveler les secrets entre les exécutions.

```python
import random
random.seed(42)
secret_de_demo = random.randint(1, 20)
print("Secret affiché pour la démonstration :", secret_de_demo)
```

Sous la version Python vérifiée pour le cours, cette démonstration affiche 4. Le jeu principal utilisera d’abord un secret fixé à 12 pour vérifier précisément les indications et le compteur.
