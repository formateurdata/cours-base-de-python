## 1. Retrouver et nettoyer du texte

Une chaîne est une séquence de caractères. Un indice commence à 0 : `nom[0]` désigne donc le premier caractère. `len(nom)` donne sa longueur. Une chaîne vide ne possède pas de caractère d’indice 0.

Les méthodes `strip()` et `lower()` renvoient respectivement une version sans espaces aux extrémités et une version en minuscules. Elles ne modifient pas la chaîne d’origine : il faut conserver le résultat si l’on souhaite le réutiliser.

```python
nom = "  Turbigo - Française  "

```

```python
nom = "  Turbigo - Française  "
nom_nettoye = nom.strip()
print("Original :", nom)
print("Nettoyé :", nom_nettoye)
print("Premier caractère nettoyé :", nom_nettoye[0])

mini = nom_nettoye.lower()
print("En minuscules :", mini)

maj = nom_nettoye.upper()
print("En majuscules :", maj)

essai1 = nom_nettoye.split("-")
print("Découpage en mots :", essai1)

essai2 = nom_nettoye.replace(" - ", " + ")
print("Remplacement :", essai2)

essai3 = nom_nettoye.replace("Française", "Italienne")
print("Remplacement :", essai3)
```

L’espace au début de la saisie disparaît après `strip()`. Nettoyer permet de comparer des noms saisis avec des espaces accidentels. L’identifiant d’une station reste préférable pour une recherche exacte.

## 2. Une liste conserve plusieurs valeurs

Une liste s’écrit entre crochets et conserve l’ordre de ses éléments. L’indice 0 correspond au premier élément ; l’indice -1 au dernier. `append()` ajoute un élément à la fin. La variable suivante contient les disponibilités de nos huit stations.

Avant d’exécuter, retrouvez dans la liste la valeur qui sera affichée en premier.

```python
valeurs_velos = [7, 14, 21, 14, 6, 22, 15, 45]

# Recupérer le nombre d'élément d'une list 
print("Nombre d'éléments :", len(valeurs_velos))

# Récupérer la valeur maximale
print("Valeur maximale :", max(valeurs_velos))

# Récupérer la valeur minimale
print("Valeur minimale :", min(valeurs_velos))

# Récupérer la somme des valeurs
print("Somme des valeurs :", sum(valeurs_velos))

# Récupérer la moyenne des valeurs
print("Moyenne des valeurs :", sum(valeurs_velos) / len(valeurs_velos))
```

```python
valeurs_velos = [7, 14, 21, 14, 6, 22, 15, 45]
# Recupérer le premier élément
print("Premier élément :", valeurs_velos[0])
# Récupérer le dernier élément
print("Dernier élément :", valeurs_velos[-1]) 
print(valeurs_velos[3:])
print(valeurs_velos[:3])
print(valeurs_velos[1:4])
```

La liste contient huit valeurs, pour un total de 144 vélos. Le minimum est 6 et le maximum 45. Ces chiffres concernent seulement notre extrait. `min()` et `max()` ne s’appliquent pas à une liste vide sans traitement particulier.

## 3. Une fiche a des champs nommés

Un dictionnaire associe des clés à des valeurs. Dans la fiche suivante, la clé `"nom"` indique ce que représente le texte. L’écriture `station["nom"]` lit ce champ. Une clé absente provoquerait une `KeyError`.

Un dictionnaire utilise des accolades ; une liste utilise des crochets. Les crochets servent aussi à lire un élément de l’une ou de l’autre structure.

```python
station = {'code': '1001', 'nom': "Quai de l'Horloge - Pont Neuf", 'velos': 7}
print(station["code"])
print(station["nom"])
print(station["velos"])
for cle, valeur in station.items():
    print(cle, ":", valeur)
```

`items()` permet de parcourir les paires clé-valeur. Les clés rendent la fiche plus explicite que trois positions dont il faudrait mémoriser le sens.

## 4. Combiner les deux structures

Une liste de dictionnaires représente plusieurs fiches ayant les mêmes champs. À chaque tour, `station` reçoit une fiche complète ; on peut alors lire ses champs.

Cette cellule fournit notre jeu de travail. Réexécutez-la pour retrouver l’état initial après une manipulation.

```python
stations = [
    {'code': '1001', 'nom': "Quai de l'Horloge - Pont Neuf", 'velos': 7},
    {'code': '1002', 'nom': 'Victoria - Place du Chatelet', 'velos': 14},
    {'code': '1003', 'nom': 'Saint-Denis - Rivoli', 'velos': 21},
    {'code': '1006', 'nom': 'Grande Truanderie - Saint-Denis', 'velos': 14},
    {'code': '1007', 'nom': 'Turbigo - Française', 'velos': 6},
    {'code': '1008', 'nom': 'Les Halles - Saint-Eustache', 'velos': 22},
    {'code': '1012', 'nom': 'Jean-Jacques Rousseau - Deux Ecus', 'velos': 15},
    {'code': '1013', 'nom': 'Place du Palais Royal', 'velos': 45},
]
for station in stations:
    print(station["code"], station["nom"], station["velos"])
```

On affiche huit lignes. Le nombre total de vélos et la recherche d’une station deviennent des traitements sur les mêmes fiches, plutôt que sur des listes séparées qui risquent de se désynchroniser.

## 5. Construire une nouvelle sélection

On peut partir d’une liste vide, parcourir les fiches et ajouter celles qui respectent une condition. L’expression `"nom" in station` testerait la présence d’une **clé** dans le dictionnaire ; elle ne chercherait pas automatiquement le texte dans ses valeurs.

Ici, le seuil est de 10 vélos.

```python
selection = []
for station in stations:
    if station["velos"] >= 10:
        selection.append(station)
print("Stations sélectionnées :", len(selection))
```

La liste contient six fiches, alors que la liste d’origine en contient toujours huit. Nous n’avons modifié aucun champ des fiches. Si l’on modifie une fiche partagée entre les deux listes, on retrouvera cette modification via les deux listes.
